#1. Bank Account Manager

```python
balance = 5000
account_holder = "Aneesh"
```

```python
def deposit(amount):
    global balance
    balance += amount
    print(f"Deposited ₹{amount}")
    print(f"Current Balance: ₹{balance}")
```

```python
def withdraw(amount):
    global balance

    if amount <= balance:
        balance -= amount
        print(f"Withdrawn ₹{amount}")
    else:
        print("Insufficient Balance")

    print(f"Current Balance: ₹{balance}")
```

```python
def check_balance():
    print(f"Balance = ₹{balance}")
```

```python
check_balance()
deposit(1000)
withdraw(3000)
withdraw(5000)
```

#2. Weather Data Analyzer

```python
temperature = [30, 32, 31, 29, 33]
humidity = [65, 70, 68, 72, 66]
```

```python
def average_temperature():
    avg = sum(temperature)/len(temperature)
    print("Average Temperature:", avg)

def average_humidity():
    avg = sum(humidity)/len(humidity)
    print("Average Humidity:", avg)
```

```python
current_temp = 32

if current_temp >= 35:
    print("Sunny")
elif current_temp >= 30:
    print("Warm")
elif current_temp >= 25:
    print("Cloudy")
else:
    print("Rainy")
```

```python
average_temperature()
average_humidity()
```
